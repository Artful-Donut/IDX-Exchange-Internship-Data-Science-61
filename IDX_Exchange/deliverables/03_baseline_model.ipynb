# Setup

```python
# Importing modules
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn import metrics
```

```python
# Importing CSVs

root = "C:/Users/donutii/Desktop/IDX-Exchange-Internship-Data-Science-61"
data_location = f"{root}/IDX_Exchange/deliverables"

testing_set = pd.read_csv(f'{data_location}/CRMLS_202605_testing_set.csv')
training_set = pd.read_csv(f'{data_location}/CRMLS_202505_202604_training_set.csv') 
```

```python
print(training_set.head())
```

```python
# Using this site as a reference https://realpython.com/linear-regression-in-python/#multiple-linear-regression-with-scikit-learn
# Splitting training set into X and Y

    # Y
target = training_set['ClosePrice']

    # X
vars = training_set.drop(columns=['ClosePrice', 'CloseDate'])

# Train a model
model = LinearRegression().fit(vars, target)

# print results
r_sq = model.score(vars, target)
print(f'R2 value: {r_sq}')
print('\n')
```

```python
print(f'Intercept: {model.intercept_}')
print(f'Coefficients:')

for c in range(len(vars.columns)):
    print(f'{vars.columns[c]}: {model.coef_[c]}')
```
