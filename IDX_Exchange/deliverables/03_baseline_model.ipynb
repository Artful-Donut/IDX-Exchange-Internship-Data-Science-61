```python
# Importing modules
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn import metrics
```

```python
# Importing CSVs

root = "C:/Users/donutii/Desktop/IDX-Exchange-Internship-Data-Science-61"
data_location = f"{root}/IDX_Exchange/deliverables"

# normal datasets
testing_set = pd.read_csv(f'{data_location}/CRMLS_202605_testing_set.csv')
training_set = pd.read_csv(f'{data_location}/CRMLS_202505_202604_training_set.csv') 

# feature engineered datasets
testing_set_fe = pd.read_csv(f'{data_location}/CRMLS_202605_testing_set_fe.csv')
training_set_fe = pd.read_csv(f'{data_location}/CRMLS_202505_202604_training_set_fe.csv') 
```

```python
print(training_set.head())
print(training_set_fe.head())
```

```python
# Using this site as a reference https://realpython.com/linear-regression-in-python/#multiple-linear-regression-with-scikit-learn
# regression
def regress_data(training, testing):

    # Splitting training set into X and Y
        # Y
    target = training['ClosePrice']
    target_testing = testing['ClosePrice']

        # X
    vars = training.drop(columns=['ClosePrice', 'CloseDate'])
    vars_testing = testing.drop(columns='ClosePrice')

    # Train a model
    model = LinearRegression().fit(vars, target)

    # print results
    r_sq = model.score(vars, target)

    # print(f'{len(vars.columns)}')
    
    # Testing results on training
    print(f'R2 value for Training: {r_sq}')
    
    # validating on testing set:
    r_sq = model.score(vars_testing, target_testing)

    print(f'R2 value on Test Set: {r_sq}')
```

```python
#print(f'Intercept: {model.intercept_}')
#print(f'Coefficients:')

#for c in range(len(vars.columns)):
#    print(f'{vars.columns[c]}: {model.coef_[c]}')

# On normal datasets
print('Regression Results for normal datasets')
regress_data(training_set, testing_set)
```

```python
print(training_set_fe.head())
print('Regression Results for feature engineered datasets')
regress_data(training_set_fe, testing_set_fe)
```
